```python
import pandas as pd

df = pd.read_csv("cleaned_car_data.csv")

print("Dataset shape:", df.shape)
print(df.head())
```

```python
categorical_columns = [
    "Fuel_Type",
    "Seller_Type",
    "Transmission"
]

df_encoded = pd.get_dummies(
    df,
    columns=categorical_columns,
    dtype=int
)

print("Encoded dataset shape:", df_encoded.shape)
print(df_encoded.columns)
```

```python
X = df_encoded.drop(
    columns=["Selling_Price", "Car_Name"]
)

Y = df_encoded["Selling_Price"]

print("X shape:", X.shape)
print("Y shape:", Y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=42
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, Y_train)

print("Random Forest training completed!")
```

```python
rf_pred = rf_model.predict(X_test)

print("First 10 Random Forest predictions:")
print(rf_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

rf_mae = mean_absolute_error(Y_test, rf_pred)
rf_mse = mean_squared_error(Y_test, rf_pred)
rf_rmse = np.sqrt(rf_mse)
rf_r2 = r2_score(Y_test, rf_pred)

print("RANDOM FOREST EVALUATION")
print("------------------------")
print("MAE :", rf_mae)
print("MSE :", rf_mse)
print("RMSE:", rf_rmse)
print("R2 Score:", rf_r2)
```

```python
comparison_models = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R2 Score"],
    "Linear Regression": [1.472892, 6.370753, 2.524035, 0.752815],
    "Random Forest": [rf_mae, rf_mse, rf_rmse, rf_r2]
})

print(comparison_models)
```

```python
feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance)
```

```python
import joblib

joblib.dump(rf_model, "car_price_random_forest.pkl")

feature_importance.to_csv(
    "car_feature_importance.csv",
    index=False
)

comparison_models.to_csv(
    "car_model_comparison.csv",
    index=False
)

print("All Day 23 files saved successfully!")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV

print("Libraries imported successfully!")
```

```python
param_grid = {
    "n_estimators": [100, 200, 300],
    "max_depth": [None, 5, 10, 15],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4]
}

print("Parameter grid created!")
```

```python
rf = RandomForestRegressor(random_state=42)

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, Y_train)

print("Hyperparameter tuning completed!")
```

```python
print("Best Parameters:")
print(grid_search.best_params_)

print("Best Cross-Validation MAE:")
print(-grid_search.best_score_)
```

```python
best_rf_model = grid_search.best_estimator_
```

```python
tuned_pred = best_rf_model.predict(X_test)
```

```python
tuned_mae = mean_absolute_error(Y_test, tuned_pred)
tuned_mse = mean_squared_error(Y_test, tuned_pred)
tuned_rmse = np.sqrt(tuned_mse)
tuned_r2 = r2_score(Y_test, tuned_pred)

print("TUNED RANDOM FOREST RESULTS")
print("---------------------------")
print("MAE :", tuned_mae)
print("MSE :", tuned_mse)
print("RMSE:", tuned_rmse)
print("R2 Score:", tuned_r2)
```

```python
import gradio as gr
import pandas as pd
```

```python
def predict_car_price(year, present_price, kms_driven,
                      owner, fuel_type, seller_type, transmission):

    new_car = pd.DataFrame([{
        "Year": year,
        "Present_Price": present_price,
        "Kms_Driven": kms_driven,
        "Owner": owner,
        "Fuel_Type_CNG": 1 if fuel_type == "CNG" else 0,
        "Fuel_Type_Diesel": 1 if fuel_type == "Diesel" else 0,
        "Fuel_Type_Petrol": 1 if fuel_type == "Petrol" else 0,
        "Seller_Type_Dealer": 1 if seller_type == "Dealer" else 0,
        "Seller_Type_Individual": 1 if seller_type == "Individual" else 0,
        "Transmission_Automatic": 1 if transmission == "Automatic" else 0,
        "Transmission_Manual": 1 if transmission == "Manual" else 0
    }])

    new_car = new_car[model.feature_names_in_]

    prediction = model.predict(new_car)[0]

    return f"Estimated Selling Price: ₹{prediction:.2f} lakhs"
```

```python
app = gr.Interface(
    fn=predict_car_price,

    inputs=[
        gr.Number(label="Manufacturing Year", value=2018),
        gr.Number(label="Present Price (₹ lakhs)", value=8.5),
        gr.Number(label="Kilometers Driven", value=30000),
        gr.Dropdown([0, 1, 2, 3], label="Previous Owners", value=0),
        gr.Dropdown(["Petrol", "Diesel", "CNG"], label="Fuel Type"),
        gr.Dropdown(["Dealer", "Individual"], label="Seller Type"),
        gr.Dropdown(["Manual", "Automatic"], label="Transmission")
    ],

    outputs=gr.Textbox(label="Predicted Car Price"),

    title="Car Selling Price Prediction",
    description="Enter the car details to estimate its selling price."
)

app.launch(share=True)
```

```python
%%writefile day26_car_price_app.py

import gradio as gr
import pandas as pd
import joblib

# Load the trained model
model = joblib.load("car_price_prediction_model.pkl")

# Prediction function
def predict_price(year, present_price, kms, owner,
                  fuel, seller, transmission):

    car = pd.DataFrame([{
        "Year": year,
        "Present_Price": present_price,
        "Kms_Driven": kms,
        "Owner": owner,
        "Fuel_Type_CNG": int(fuel == "CNG"),
        "Fuel_Type_Diesel": int(fuel == "Diesel"),
        "Fuel_Type_Petrol": int(fuel == "Petrol"),
        "Seller_Type_Dealer": int(seller == "Dealer"),
        "Seller_Type_Individual": int(seller == "Individual"),
        "Transmission_Automatic": int(transmission == "Automatic"),
        "Transmission_Manual": int(transmission == "Manual")
    }])

    car = car[model.feature_names_in_]
    result = model.predict(car)[0]

    return f"Estimated Selling Price: ₹{result:.2f} lakhs"


# Create Gradio interface
app = gr.Interface(
    fn=predict_price,
    inputs=[
        gr.Number(label="Year", value=2018),
        gr.Number(label="Present Price", value=8.5),
        gr.Number(label="Kilometers Driven", value=30000),
        gr.Dropdown([0, 1, 2, 3], value=0, label="Previous Owners"),
        gr.Dropdown(["Petrol", "Diesel", "CNG"], value="Petrol", label="Fuel"),
        gr.Dropdown(["Dealer", "Individual"], value="Individual", label="Seller"),
        gr.Dropdown(["Manual", "Automatic"], value="Manual", label="Transmission")
    ],
    outputs=gr.Textbox(label="Predicted Price"),
    title="Car Selling Price Prediction"
)

app.launch(share=True)
```
